# Patient Management System

```python
class Patient:
    def __init__(self, patient_id, name, age, gender, disease):
        self.patient_id = patient_id
        self.name = name
        self.age = age
        self.gender = gender
        self.disease = disease
        self.admission_status = True
        self.assigned_doctor = None

    def display_info(self):
        print(f"""
        Patient ID: {self.patient_id}
        Name: {self.name.capitalize()}
        Age: {self.age}
        Gender: {self.gender.capitalize()}
        Disease: {self.disease.capitalize()}
        Admission Status: {"Admitted" if self.admission_status else "Discharged"}
        Assigned Doctor: {self.assigned_doctor.name.title() if self.assigned_doctor else "Not assigned"}
        """)

    def admit(self):
        self.admission_status = True

    def discharge(self):
        self.admission_status = False

    def assign_doctor(self,doctor):
        self.assigned_doctor = doctor
```

```python
class Doctor:
    def __init__(self, doctor_id, name, specialization):
        self.doctor_id = doctor_id
        self.name = name
        self.specialization = specialization

    def display_info(self):
        info = f"""ID: {self.doctor_id}, 
Name: {self.name.title()}, 
Specialization: {self.specialization.capitalize()}"""
        print(info)
```

```python
class Hospital:
    def __init__(self):
        self.patients = []
        self.doctors = []

    def add_patients(self, patient):
        self.patients.append(patient)
        print(f"Patient registered with ID : {patient.patient_id}")

    def add_doctor(self, doctor):
        self.doctors.append(doctor)

    def view_doctors(self):
        for doctor in self.doctors:
            doctor.display_info()
        
    def view_patients(self):
        for patient in self.patients:
            patient.display_info()

    def discharge_patient(self, patient_id):
        patient = self.search_patient(patient_id)
        if patient:
            patient.discharge()
            print(f"Patient {patient.name.capitalize()} is discharged.")
        else:
            print("Patient not found!")

    def search_patient(self, patient_id):
        for patient in self.patients:
            if patient.patient_id == patient_id:
                return patient
        return None

    def assigned_doctor_to_patient(self, patient_id, doctor):
        patient = self.search_patient(patient_id)
        if patient:
            patient.assign_doctor(doctor)
            print(f"Doctor {doctor.name.title()}({doctor.specialization.capitalize()}) assigned to {patient.name.capitalize()}.")
        else:
            print("Patient not found.")

    def view_reports_by_doctor(self, doctor_id):
        doctor = next((doc for doc in self.doctors if doc.doctor_id == doctor_id),None)
        if not doctor:
            print(f"Doctor with ID '{doctor_id}' not found!")
            return 
        filtered = [p for p in self.patients if p.assigned_doctor == doctor]
        if filtered:
            print(f"\nPatients assigned to {doctor.name.title()}")
            for p in self.patients:
                p.display_info()
        else:
            print(f"No patients are assigned to {doctor.name.title()}")
```

```python
def main():
    hospital = Hospital()
    doctor_base = "D"
    doctor_id_counter = 1
    patient_id_counter = 1001
    
    while True:
        menu = {1:"Register New Patient", 
                2:"View All Patients", 
                3:"Add New Doctor",
                4:"View All Doctors",
                5:"Assign Doctor to Patient",
                6:"Discharge Patient", 
                7:"Search Patient by ID", 
                8:"View Reports by Doctor",
                9:"Exit" }
        
        print("\n======= Patient Management System =======")
        for key,value in menu.items():
            print(f"{key}.{value}")
        
        choice = int(input("\nEnter your choice: "))

        if choice == 1:
            print("------------------------------------")
            name = input("Enter patient name: ")
            age = int(input("Enter age: "))
            gender = input("Enter gender: ")
            disease = input("Enter disease: ")
            patient = Patient(patient_id_counter, name, age, gender, disease)
            hospital.add_patients(patient)
            patient_id_counter += 1

        elif choice == 2:
            print("------------------------------------")
            hospital.view_patients()

        elif choice == 3:
            print("------------------------------------")
            doctor_id = doctor_base + str(doctor_id_counter)
            name = input("Enter doctor name: ")
            specialization = input("Enter specialization: ")
            doctor = Doctor(doctor_id, name, specialization)
            hospital.add_doctor(doctor)
            doctor_id_counter += 1

        elif choice ==4:
            print("------------------------------------")
            hospital.view_doctors()
        
        elif choice == 5:
            print("------------------------------------")
            try:
                patient_id = int(input("Enter patient ID: "))
            except ValueError:
                print("Invalid patient ID.")
            print("Assigning Doctor to patient ID ",patient_id)
            doctor_id = input("Enter doctor ID: ")
            doctor = next((doc for doc in hospital.doctors if doc.doctor_id == doctor_id),None)
            if doctor:
                hospital.assigned_doctor_to_patient(patient_id, doctor)
            else:
                print("Doctor not found!")

        elif choice == 6:
            print("------------------------------------")
            try:
                patient_id = int(input("Enter patient ID: "))
            except ValueError:
                print("Invalid patient ID.")
            hospital.discharge_patient(patient_id)

        elif choice == 7:
            print("------------------------------------")
            try:
                patient_id = int(input("Enter patient ID: "))
            except ValueError:
                print("Invalid patient ID.")
            patient = hospital.search_patient(patient_id)
            if patient:
                patient.display_info()
            else:
                print("Patient not found!")

        elif choice == 8:
            print("------------------------------------")
            try:
                doctor_id = input("Enter doctor ID: ")
            except ValueError:
                print("Invalid doctor ID.")
            hospital.view_reports_by_doctor(doctor_id)
            
        elif choice == 9:
            print("Exiting the system!")
            break

        else:
            print("Invalid choice! Please try again!")

if __name__ == "__main__":
    main()
```
